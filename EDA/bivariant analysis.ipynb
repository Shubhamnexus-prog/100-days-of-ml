```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

tips = sns.load_dataset("tips")

```

```python
tips
```

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Total Bill vs Tip")
plt.show()
```

```python
# correlation

tips[['total_bill','tip']].corr()
```

```python
# box plot--

sns.boxplot(x='sex',y='total_bill',data=tips)
plt.show()
```

```python
# correlation heatmap

corr=tips[['total_bill','tip','size']].corr()
plt.figure(figsize=(5,4))
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)
plt.show()
```

```python
# violin plot

plt.figure(figsize=(8,5))

sns.violinplot(
    x='sex',
    y='total_bill',
    data=tips
)
plt.show()
```

```python
# strip plot

plt.figure(figsize=(8,5))
sns.stripplot(
    x='day',
    y='tip',
    data=tips
)
plt.show()
```

```python
# swarm plot

plt.figure(figsize=(8,5))
sns.swarmplot(
    x='day',
    y='tip',
    data=tips
)
plt.show()
```

```python
# box plot with hue

plt.figure(figsize=(8,5))

sns.boxplot(
    x='day',
    y='tip',
    hue='sex',
    data=tips
)

plt.show()
```

```python
pd.crosstab(
    tips['sex'],
    tips['smoker']
)
```

```python
# pair plot

sns.pairplot(tips)

plt.show()
```

```python
sns.pairplot(
    tips,
    hue='sex'
)

plt.show()
```

```python
# group by analysis--

tips.groupby('sex')['tip'].mean()
```

```python
# average

tips.groupby('day')['total_bill'].mean()
```

```python
tips.groupby('smoker')['tip'].mean()
```

```python
tips.groupby('time')['total_bill'].mean()
```

```python
# pivot table--
tips.pivot_table(
    values='tip',
    index='day',
    columns='sex',
    aggfunc='mean'
)
```

```python
pivot = tips.pivot_table(
    values='tip',
    index='day',
    columns='sex',
    aggfunc='mean'
)

plt.figure(figsize=(6,4))

sns.heatmap(
    pivot,
    annot=True,
    cmap='YlGnBu'
)

plt.show()
```
