```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset=pd.read_csv(r'C:\Users\SHUBHAM\Downloads\wine_data1.csv')
dataset
```

```python
print("Dataset Shape:", dataset.shape)
display(dataset.head())
```

```python
print("\nDataset Information:")
print(dataset.info())

print("\nMissing Values:")
print(dataset.isnull().sum())
```

```python
features = [
    'fixed acidity',
    'volatile acidity',
    'citric acid',
    'residual sugar',
    'chlorides',
    'free sulfur dioxide',
    'total sulfur dioxide',
    'density',
    'pH',
    'sulphates',
    'alcohol'
]
```

```python
X = dataset[features].copy()

print("\nOriginal Features:")
display(X.head())
```

```python
print("\nOriginal Data Statistics:")
display(X.describe())
```

```python

plt.figure(figsize=(14, 6))

sns.boxplot(data=X)

plt.title("Before Scaling - Original Data")
plt.xticks(rotation=45)
plt.ylabel("Value")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    X['fixed acidity'],
    kde=True
)

plt.title("Before Scaling - Fixed Acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Frequency")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X = dataset.drop(['quality', 'wine_type'], axis=1)
y = dataset['quality']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=0
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
np.round(X_train.describe(), 1)
```

```python
fig, (ax1, ax2) = plt.subplots(
    ncols=2,
    figsize=(12, 5)
)

# Before Scaling
ax1.scatter(
    X_train['fixed acidity'],
    X_train['volatile acidity'],
    c=y_train
)

ax1.set_title("Before Scaling")
ax1.set_xlabel("Fixed Acidity")
ax1.set_ylabel("Volatile Acidity")


# After Scaling
ax2.scatter(
    X_train['fixed acidity'],
    X_train['volatile acidity'],
    c=y_train
)

ax2.set_title("After Scaling")
ax2.set_xlabel("Fixed Acidity")
ax2.set_ylabel("Volatile Acidity")

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(
    ncols=2,
    figsize=(12, 5)
)

ax1.scatter(
    X_train['alcohol'],
    X_train['fixed acidity'],
    c=y_train
)

ax1.set_title("Before Scaling")
ax1.set_xlabel("Alcohol")
ax1.set_ylabel("Fixed Acidity")

ax2.scatter(
    X_train['alcohol'],
    X_train['fixed acidity'],
    c=y_train
)

ax2.set_title("After Standard Scaling")
ax2.set_xlabel("Alcohol")
ax2.set_ylabel("Fixed Acidity")

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)

display(X_train_scaled.head())
```

```python
fig, (ax1, ax2) = plt.subplots(
    ncols=2,
    figsize=(14, 5)
)

# Before Scaling
ax1.scatter(
    X_train['alcohol'],
    X_train['fixed acidity'],
    c=y_train
)

ax1.set_title("Before Scaling")
ax1.set_xlabel("Alcohol")
ax1.set_ylabel("Fixed Acidity")


# After Scaling
ax2.scatter(
    X_train_scaled['alcohol'],
    X_train_scaled['fixed acidity'],
    c=y_train
)

ax2.set_title("After Standard Scaling")
ax2.set_xlabel("Alcohol (Scaled)")
ax2.set_ylabel("Fixed Acidity (Scaled)")

plt.tight_layout()
plt.show()
```
