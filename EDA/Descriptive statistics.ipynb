```python
#What is Descriptive Statistics?
#Mean
#Median
#Mode
#Minimum & Maximum
#Range
#Quartiles (Q1, Q2, Q3)
#Percentiles
#Variance
#Standard Deviation
#Skewness
#Kurtosis
```

```python
import seaborn as sns

df = sns.load_dataset("titanic")
```

```python
df
```

```python
df.describe()
```

```python
# mean-

df['age'].mean()
```

```python
#median--

df['age'].median()
```

```python
# mode

df['embarked'].mode()
```

```python
# first mode--

df['embarked'].mode(0)
```

```python
# min value--

df['age'].min()
```

```python
# max value--

df['age'].max()
```

```python
# Range = Maximum − Minimum

df['age'].max() - df['age'].min()
```

```python
# count

df['age'].count()
```

```python
# sum--

df['fare'].sum()
```

```python
# variance

df['age'].var()
```

```python
# standard deviation  

df['age'].std()
```

```python

# quartiles

df['age'].quantile(0.25)
```

```python
df['age'].quantile(0.75)
```

```python
# percentile

df['age'].quantile(0.90)
```

```python
df['age'].quantile(0.95)
```

```python
# skewness

df['age'].skew()
```

```python
# kurtosis

df['age'].kurt()
```

```python
# aggregate multiple statostics

df['age'].agg(['count', 'mean', 'median', 'min', 'max', 'std'])
```

```python
# describe all columns

df.describe(include='all')
```
