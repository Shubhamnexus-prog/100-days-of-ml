```python
import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

plt.pie(marks)

plt.show()
```

```python
# add label--

subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

plt.pie(marks,labels=subjects)

plt.show()
```

```python
#show percentages--

subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

plt.pie(marks,labels=subjects,autopct= "%1.1f%%")

plt.show()
```

```python
#change color==

subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]
colors=['red','blue','green','orange']

plt.pie(marks,labels=subjects,colors=colors,autopct="%1.1f%%")

plt.show()
```

```python


subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

colors = ["red", "blue", "green", "orange"]

plt.pie(
    marks,
    labels=subjects,
    colors=colors,
    autopct="%1.1f%%"
)

plt.show()
plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90
)
plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90
)
plt.show()
```

```python
#. Explode a Slice

subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

explode = [0, 0, 0, 0.2]

plt.pie(
    marks,
    labels=subjects,
    explode=explode,
    autopct="%1.1f%%"
)

plt.show()
```

```python
# add shadow--

subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

explode = [0, 0, 0, 0.2]

plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    shadow=True
)

plt.show()
```

**Donut chart**

```python


subjects = ["Math", "Science", "English", "Computer"]
marks = [80, 70, 60, 90]

plt.figure(figsize=(8,8))

plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%"
)

centre_circle = plt.Circle((0,0), 0.70, color="white")

fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Donut Chart")

plt.show()
```
