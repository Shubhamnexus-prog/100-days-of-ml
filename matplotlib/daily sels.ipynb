```python
#full example


months = ["Jan","Feb","Mar","Apr","May","Jun",
          "Jul","Aug","Sep","Oct","Nov","Dec"]

sales = [100,200,300,400,500,600,
         700,800,900,1000,1100,1200]

plt.figure(figsize=(12,6))

plt.plot(
    months,
    sales,
    color="red",
    linewidth=3,
    linestyle="--",
    marker="o",
    markersize=8
)

plt.title("Monthly Sales Report")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
#Weekly Temperature

temp=[25,27,40,31,37,38,39]
days=['mon','tue','wed','thr','fri','sat','sun']
plt.figure(figsize=(10,5))

plt.plot(
    temp,
    days,
    color='green',
    marker='s',
)
plt.title("Weekly Temperature")
plt.xlabel('temp')
plt.ylabel('days')
plt.grid(True)
plt.show()
    
```
