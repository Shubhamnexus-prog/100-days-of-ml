**bar charts**

```python
import matplotlib.pyplot as plt

students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.bar(students, marks)

plt.show()
```

```python
#add title and label

students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel('student')
plt.ylabel('marks')

plt.show()
```

```python
# change bar color


students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.bar(students, marks,color='green')

plt.show()
```

```python
#bar width



students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.bar(students, marks,width=0.5)

plt.show()
```

```python
# add edge color



students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.bar(students, marks ,color='skyblue',edgecolor='black')

plt.show()
```

```python


students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.figure(figsize=(8,5))

plt.bar(
    students,
    marks,
    color="orange",
    edgecolor="black",
    width=0.6
)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
```

**Horizontal bar chart**

```python

students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

plt.barh(students, marks, color="green")

plt.title("Student Marks")
plt.xlabel("Marks")
plt.ylabel("Students")

plt.show()
```

```python
# add value top of bar
students = ["Rahul", "Priya", "Aman", "Neha"]
marks = [85, 92, 78, 88]

bars = plt.bar(students, marks, color="skyblue")

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        str(bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
#Requirements:

#Figure size (10,5)
#Blue bars
#Black edge color
#Width 0.5
#Title: Monthly Sales
#X label: Months
#Y label: Sales
#Grid only on the Y-axis

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 170, 210, 230]

plt.figure(figsize=(10,5))
plt.bar(months,sales,width=0.5,color='blue',edgecolor='black')
plt.title('Monthly Sales')
plt.xlabel('months')
plt.ylabel('sales')
plt.grid(sales)
plt.show()          
```

```python


# Data
departments = ["HR", "IT", "Sales", "Finance", "Marketing"]
employees = [15, 40, 30, 20, 25]

# Figure Size
plt.figure(figsize=(10, 6))

# Create Bar Chart
bars = plt.bar(
    departments,
    employees,
    color="orange",
    edgecolor="black"
)

# Add Employee Count on Top of Each Bar
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(bar.get_height()),
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

# Title and Labels
plt.title("Employees by Department", fontsize=16, fontweight="bold")
plt.xlabel("Department", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

# Grid on Y-axis
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Display the Graph
plt.show()
```
