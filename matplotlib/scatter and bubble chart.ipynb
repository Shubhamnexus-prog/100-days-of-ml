**Scatter plot**

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.scatter(x, y)

plt.show()
```

```python
# add title and label

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.scatter(x, y)

plt .title("Scatter plot example")
plt.xlabel("x value")
plt.ylabel("y value")

plt.show()
```

```python
#change marker color--

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.scatter(x, y)
plt.scatter(x,y,color='red')

plt.show()
```

```python
#change a shape


x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.scatter(x, y)

plt.scatter(x,y,marker='s')
plt.show()
```

```python
# size--

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.scatter(x, y)

plt.scatter(x,y,s=100)
plt.show()
```

```python
#transparency--(alpha)

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.scatter(x, y)

plt.scatter(x,y,color='blue',s=100,alpha=0.5)
plt.show()
```

**Bubble chart**

```python

students = ["A", "B", "C", "D", "E"]

hours = [2, 4, 6, 8, 10]
marks = [40, 60, 70, 85, 95]

size = [100, 200, 300, 400, 500]

plt.figure(figsize=(8,5))

plt.scatter(
    hours,
    marks,
    s=size,
    color="orange",
    alpha=0.6
)

plt.title("Bubble Chart")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)

plt.show()
```

```python


height = [150,155,160,165,170,175,180]
weight = [45,50,55,60,68,75,82]

plt.figure(figsize=(8,5))

plt.scatter(
    height,
    weight,
    color="purple",
    s=100,
    marker="D"
)

plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.grid(True)

plt.show()
```

```python
#Requirements:

#Figure size (8,5)
#Blue markers
#Circle marker
#Size 100
#Grid
#Title: Age vs Salary
#X-axis: Age
#Y-axis: Salary

age = [18, 20, 22, 24, 26, 28]
salary = [20000, 25000, 30000, 40000, 50000, 60000]
plt.figure(figsize=(8,5))
plt.scatter(age,salary,color='blue',marker='o',s=100)
plt.title("Age vs Salary")
plt.xlabel('Age')
plt.ylabel('Salary')
plt.grid(True)
plt.show()
```

```python
# project--

advertising = [10,20,30,40,50,60,70]
sales = [100,150,180,220,260,300,350]

plt.figure(figsize=(10,6))
plt.scatter(advertising,sales,color='green',marker='d',s=150,alpha=0.8)
plt.title("Advertising Budget vs Sales")
plt.xlabel("Advertising Budget (Lakhs) ")
plt.ylabel("Sales(Lakhs)")
plt.grid(True)
plt.show()
```
