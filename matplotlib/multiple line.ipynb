```python
import matplotlib.pyplot as plt
```

```python
#multiple line---
x = [1, 2, 3, 4, 5]

math = [70, 75, 80, 85, 90]
science = [65, 70, 78, 82, 88]

plt.plot(x, math)
plt.plot(x, science)

plt.show()
```

```python
# legend


x = [1, 2, 3, 4, 5]

math = [70, 75, 80, 85, 90]
science = [65, 70, 78, 82, 88]

plt.plot(x, math, label="Math")
plt.plot(x, science, label="Science")

plt.legend()

plt.show()
```

```python
#color
x = [1, 2, 3, 4, 5]

math = [70, 75, 80, 85, 90]
science = [65, 70, 78, 82, 88]

plt.plot(x, math, label="Math",color='green')
plt.plot(x, science, label="Science",color='red')

plt.legend()

plt.show()
```

```python

#markers

x = [1, 2, 3, 4, 5]

math = [70, 75, 80, 85, 90]
science = [65, 70, 78, 82, 88]

plt.plot(x, math, label="Math",marker='v')
plt.plot(x, science, label="Science",marker='d')

plt.legend()

plt.show()
```

```python
#line style

x = [1, 2, 3, 4, 5]

math = [70, 75, 80, 85, 90]
science = [65, 70, 78, 82, 88]

plt.plot(x, math, label="Math",linestyle=':')
plt.plot(x, science, label="Science",linestyle='--')

plt.legend()

plt.show()
```

```python

#Requirements:

#Morning → Blue, circle marker
#Evening → Red, square marker
#Grid

#Legend
#Figure size (10,5)


days = ["Mon", "Tue", "Wed", "Thu", "Fri","sat","sun"]

morning = [20, 22, 25, 24, 23,21,19]
evening = [28, 30, 31, 29, 27,31,33]
plt.figure(figsize=(10,5))
plt.plot(days,morning,label='morning', color='blue',marker='o')
plt.plot(days,evening,label='evening',color='red',marker='s')
plt.grid(True)
plt.legend()

plt.show()
```

```python
#Requirements:

#Profit → Green line
#Loss → Orange dashed line
#Legend
#Grid
#Title

months = ["Jan", "Feb", "Mar", "Apr", "May"]

profit = [10, 15, 18, 22, 25]
loss = [5, 6, 4, 3, 2]

plt.plot(months,profit,label='profit',color='green')
plt.plot(months,loss,label='loss',color='orange')

plt.title("Monthly profit-loss dasbord")

plt.grid(True)
plt.legend()
plt.show()
```

```python

#project--

#student performance comparison--

#Requirements:

#Rahul → Blue circle markers
#Priya → Red square markers
#Line width 3
#Figure size (12,6)
#Title: Student Performance Comparison
#X-axis: Subjects
#Y-axis: Marks
#Grid
#Legend

subjects = ["Math", "Science", "English", "History", "Computer"]

Rahul = [80, 75, 85, 78, 90]
Priya = [82, 79, 88, 80, 95]

plt.figure(figsize=(12,6))

plt.plot(subjects,Rahul,label='Rabul',color='blue', marker='o',linewidth=3)
plt.plot(subjects,Priya,label='Priya',color='red', marker='s',linewidth=3)

plt.title("Student Performance Comparison")

plt.xlabel('subjects')
plt.ylabel('marks')

plt.grid(True)

plt.legend()

plt.show()













```
