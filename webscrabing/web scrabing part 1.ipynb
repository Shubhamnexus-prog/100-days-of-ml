```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/Python_(programming_language)"

headers = {
    "User-Agent": "MyWikipediaScraper/1.0 (student project; your_email@example.com)"
}

response = requests.get(url, headers=headers)

print(response.status_code)

soup = BeautifulSoup(response.text, "lxml")

print(soup.title.text)
```

```python
# Infobox table find karo
infobox = soup.find("table", class_="infobox")

data = {}

if infobox:
    rows = infobox.find_all("tr")

    for row in rows:
        heading = row.find("th")
        value = row.find("td")

        if heading and value:
            data[heading.text.strip()] = value.text.strip()

# Dictionary print karo
data
```

```python
import pandas as pd

df = pd.DataFrame(
    list(data.items()),
    columns=["Feature", "Information"]
)

df
```

```python
df["Information"] = (
    df["Information"]
    .str.replace("\xa0", " ", regex=False)
    .str.replace("\n", " ", regex=False)
    .str.strip()
)

df
```

```python
df.to_csv(
    "python_infobox.csv",
    index=False
)

print("CSV Saved Successfully")
```

```python
api_url = "https://en.wikipedia.org/w/api.php"

params = {
    "action": "query",
    "format": "json",
    "list": "categorymembers",
    "cmtitle": "Category:Countries",
    "cmlimit": 100
}


response = requests.get(
    api_url,
    params=params,
    headers=headers
)

result = response.json()

result.keys()
```

```python
countries = []

for item in result["query"]["categorymembers"]:
    countries.append(item["title"])


print(len(countries))
print(countries[:20])
```

```python
def scrape_country(country):

    url = "https://en.wikipedia.org/wiki/" + country.replace(" ", "_")

    response = requests.get(
        url,
        headers=headers
    )


    soup = BeautifulSoup(
        response.text,
        "lxml"
    )


    data = {}

    data["Country"] = country


    infobox = soup.find(
        "table",
        class_="infobox"
    )


    if infobox:

        rows = infobox.find_all("tr")


        for row in rows:

            key = row.find("th")
            value = row.find("td")


            if key and value:

                data[key.text.strip()] = (
                    value.text
                    .replace("\n"," ")
                    .strip()
                )


    return data
```
